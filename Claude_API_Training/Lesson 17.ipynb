## Lesson Overview

**What this lesson teaches:** how prompt caching reuses a stable prompt prefix so repeated Claude API requests can process fewer input tokens.

**What's happening under the hood:**
1. Build a large, stable prefix from tool definitions and a system prompt.
2. Mark the end of each reusable prefix with `cache_control`.
3. Make an initial request that writes eligible tokens to the cache.
4. Repeat the request with a different user question and read the cached prefix.
5. Inspect usage fields to distinguish cache creation, cache reads, and uncached input.

**You will build:** a reusable chat helper that automatically caches tools and system instructions, followed by a two-request experiment.

> Prompt caching changes input processing and billing; it does not store or reuse Claude's answer. Live examples are opt-in because API calls may incur a charge.

# Lesson 17: Prompt Caching

Prompt caching is useful when many requests begin with the same long context: policies, examples, tool definitions, documents, or coding instructions. Claude still generates a fresh response to every question, but an unchanged cached prefix can be reused.

```text
tools + long system prompt + cache breakpoint | changing user message
                 stable prefix             | dynamic suffix
```

Caching only helps when the prefix is large enough for the selected model and remains byte-for-byte stable. The default ephemeral cache lifetime is five minutes and refreshes when the cache is read.

## Cache-Aware Helpers

A cache breakpoint marks the end of a reusable prefix. The helper below places one on the final tool and another on the system text. Because request order is tools → system → messages, these breakpoints let Claude reuse the stable tools and long instructions while user messages continue to change.

Up to four cache breakpoints may be used in one request. Keep frequently changing material after the last breakpoint.

````python
def add_user_message(messages, message):
    content = message.content if hasattr(message, 'content') else message
    messages.append({'role': 'user', 'content': content})


def add_assistant_message(messages, message):
    content = message.content if hasattr(message, 'content') else message
    messages.append({'role': 'assistant', 'content': content})


def cache_tools(tool_definitions):
    cached = [dict(tool) for tool in tool_definitions]
    if cached:
        cached[-1]['cache_control'] = {'type': 'ephemeral'}
    return cached


def cache_system(system_text):
    return [
        {
            'type': 'text',
            'text': system_text,
            'cache_control': {'type': 'ephemeral'},
        }
    ]


def chat(messages, system=None, tools=None, max_tokens=800):
    if client is None:
        raise RuntimeError('Complete the API setup before making a request.')

    params = {
        'model': model,
        'max_tokens': max_tokens,
        'messages': messages,
    }
    if tools:
        params['tools'] = cache_tools(tools)
    if system:
        params['system'] = cache_system(system)

    return client.messages.create(**params)


def text_from_message(message):
    return '\n'.join(
        block.text for block in message.content if block.type == 'text'
    )


def cache_usage(message):
    usage = message.usage
    return {
        'uncached_input_tokens': usage.input_tokens,
        'cache_creation_input_tokens': getattr(usage, 'cache_creation_input_tokens', 0) or 0,
        'cache_read_input_tokens': getattr(usage, 'cache_read_input_tokens', 0) or 0,
        'output_tokens': usage.output_tokens,
    }
````

## Compare the Measurements

`input_tokens` represents input after the final cache breakpoint that was not read from or written to the cache. The two cache fields account for the reusable prefix. Do not judge caching from wall-clock time alone: network conditions and output length also affect latency.

## Troubleshooting Cache Misses

If `cache_read_input_tokens` is zero, check these causes:

1. **Prefix too short:** the content before the breakpoint must meet the selected model's minimum cacheable length.
2. **Prefix changed:** tools, system blocks, images, documents, and messages before the breakpoint must match in content and order.
3. **Entry expired:** the default ephemeral lifetime is five minutes; run the second request promptly.
4. **Different model:** cache entries are model-specific.
5. **Breakpoint moved:** changing where `cache_control` appears changes the cached prefix.
6. **Concurrent first requests:** wait for the cache-writing request to begin before expecting another request to read its new entry.

For a longer reuse window, supported API versions can use `{'type': 'ephemeral', 'ttl': '1h'}`. Cache writes, five-minute writes, one-hour writes, and reads have different input-token rates, so use the response usage fields with the current model pricing rather than hard-coding cost assumptions.

Try these one at a time and predict whether the second request will hit the cache:

1. **Change only the user question:** confirm that the system prefix is still read.
2. **Change one character in `code_prompt`:** observe a new cache write instead of the prior full-prefix hit.
3. **Reorder two tools:** compare the cache usage fields.
4. **Move dynamic context:** place a timestamp after the breakpoint and keep the stable prefix unchanged.
5. **Test conversation caching:** append earlier messages and put a breakpoint on a stable content block near the end of the history.
6. **Estimate savings:** combine measured write/read token counts with the current pricing for your model and expected number of repeated requests.

## Summary

- Prompt caching reuses a matching input prefix; it never reuses the generated answer.
- Put stable, sufficiently long content first and dynamic content afterward.
- Add `cache_control` at deliberate prefix boundaries, with no more than four breakpoints.
- Expect the first request to write eligible tokens and later matching requests to read them.
- Verify behavior with `cache_creation_input_tokens` and `cache_read_input_tokens`.
- Cache lifetime, write premiums, and read discounts matter when estimating whether caching is worthwhile.
